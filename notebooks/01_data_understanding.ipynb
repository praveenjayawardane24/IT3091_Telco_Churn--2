##### Cell 1 — Import and load the dataset

```python

# ==========================================================
# IT3091 - Telco Customer Churn
# Stage 2: Dataset Understanding
# Responsibility: IT24102178 - Data Understanding
# ==========================================================

from pathlib import Path
import pandas as pd
import numpy as np

# Find project root (works from project root or notebooks folder)
current_dir = Path.cwd()
project_root = (
    current_dir.parent
    if current_dir.name == "notebooks"
    else current_dir
)

dataset_path = (
    project_root
    / "data"
    / "raw"
    / "WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

df = pd.read_csv(dataset_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

##### Cell 3 — Examine column names and data types

```python

# Responsibility: IT24102178 - Data Understanding

print("COLUMN INFORMATION")
print("-" * 50)

df.info()

print("\nCOLUMN DATA TYPES")
display(
    df.dtypes.reset_index().rename(
        columns={"index": "Column", 0: "Data Type"}
    )
)
```

##### Cell 4 — Check missing values

```python

# Responsibility: IT24102310 - Data Quality Assessment

print("MISSING VALUES PER COLUMN")

missing_summary = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": (df.isna().mean() * 100).round(2)
})

display(missing_summary)

print("\nBlank or whitespace-only TotalCharges:")
print(
    df["TotalCharges"]
    .astype("string")
    .str.strip()
    .eq("")
    .sum()
)
```

##### Cell 5 — Check duplicate records and customer IDs

```python

# Responsibility: IT24102310 - Data Quality Assessment

print("Duplicate full rows:", df.duplicated().sum())

print(
    "Duplicate customer IDs:",
    df["customerID"].duplicated().sum()
)

print(
    "Unique customer IDs:",
    df["customerID"].nunique()
)
```

##### Cell 6 — Examine the churn target distribution

```python

# Responsibility: IT24102178 - Data Understanding
# Validation relevance: IT24101808

print("CHURN CLASS DISTRIBUTION")

churn_counts = df["Churn"].value_counts(dropna=False)
churn_percentages = (
    df["Churn"].value_counts(
        normalize=True,
        dropna=False
    ) * 100
).round(2)

churn_summary = pd.DataFrame({
    "Count": churn_counts,
    "Percentage": churn_percentages
})

display(churn_summary)
```

##### Cell 7 — Numerical summary

```python

# Responsibility: IT24102178 - Data Understanding

print("NUMERICAL COLUMN SUMMARY")

display(df.describe().T)
```

##### Cell 8 — Investigate blank TotalCharges records

```python

# ==========================================================
# Responsibility: IT24102310 - Data Quality Assessment
# Investigate blank TotalCharges values
# ==========================================================

blank_mask = (
    df["TotalCharges"]
    .astype("string")
    .str.strip()
    .eq("")
)

blank_records = df.loc[
    blank_mask,
    [
        "customerID",
        "tenure",
        "MonthlyCharges",
        "TotalCharges",
        "Contract",
        "Churn"
    ]
]

print("Number of blank TotalCharges records:", len(blank_records))

display(blank_records)

print("\nTenure distribution for blank records:")
print(blank_records["tenure"].value_counts())
```

##### Cell 9 — Check whether TotalCharges contains other invalid values

```python

# Responsibility: IT24102310 - Data Quality Assessment

total_charges_numeric = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

invalid_mask = total_charges_numeric.isna()

print("Total invalid TotalCharges values:", invalid_mask.sum())

print(
    "Invalid values other than blanks:",
    (invalid_mask & ~blank_mask).sum()
)

print("\nTotalCharges numerical summary:")
display(total_charges_numeric.describe())
```

##### Cell 10 — Examine categorical feature values

```python

# Responsibility: IT24102310 - Data Quality Assessment

categorical_columns = df.select_dtypes(
    include=["str", "object", "category"]
).columns.drop(["customerID", "TotalCharges"])

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False).to_string())
```

##### Cell 11 — Create a cleaned working copy

```python

# ==========================================================
# Responsibility: IT24102310 - Data Quality Assessment
# Initial data type correction
# ==========================================================

df_clean = df.copy()

# Convert TotalCharges from string to numeric
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

# Replace missing TotalCharges with zero ONLY for tenure = 0
zero_tenure_mask = (
    df_clean["tenure"].eq(0)
    & df_clean["TotalCharges"].isna()
)

df_clean.loc[zero_tenure_mask, "TotalCharges"] = 0.0

print("Replaced values:", zero_tenure_mask.sum())
print("Remaining missing TotalCharges:",
      df_clean["TotalCharges"].isna().sum())

print("TotalCharges data type:",
      df_clean["TotalCharges"].dtype)

print("Dataset shape:", df_clean.shape)
```

##### Cell 12 — Verify categorical values in a compact table

```python

# Responsibility: IT24102310 - Data Quality Assessment

categorical_cols = df_clean.select_dtypes(
    include=["str", "object", "category"]
).columns.drop(["customerID", "Churn"])

category_summary = pd.DataFrame({
    "Feature": categorical_cols,
    "Unique Values": [
        df_clean[col].nunique(dropna=False)
        for col in categorical_cols
    ],
    "Categories": [
        ", ".join(
            df_clean[col]
            .astype(str)
            .unique()
        )
        for col in categorical_cols
    ]
})

display(category_summary)
```

##### Cell 13 — Check the cleaned dataset for remaining problems

```python

# Responsibility: IT24102310 - Data Quality Assessment

print("FINAL INITIAL QUALITY CHECK")
print("-" * 45)

print("Rows:", len(df_clean))
print("Columns:", df_clean.shape[1])
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate customer IDs:",
      df_clean["customerID"].duplicated().sum())

print("\nNumeric feature ranges:")
display(
    df_clean[
        ["tenure", "MonthlyCharges", "TotalCharges"]
    ].agg(["min", "max", "mean"]).round(2)
)

print("\nTarget categories:")
print(df_clean["Churn"].value_counts())
```
