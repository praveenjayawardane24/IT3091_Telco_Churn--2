##### Cell 1 — Import shared functions and load data


```python

# ==========================================================
# IT3091 - Shared Evaluation Verification
# ==========================================================

import sys
from pathlib import Path

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from sklearn.dummy import DummyClassifier

from src.data_preparation import (
    load_and_clean_data,
    separate_features_target
)

from src.validation import create_train_test_split
from src.imbalance import build_model_pipeline
from src.evaluation import evaluate_cv

df = load_and_clean_data(project_root)

X, y = separate_features_target(df)

X_train, X_test, y_train, y_test = (
    create_train_test_split(X, y)
)

print("Evaluation imports successful!")
print("Training shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

##### Cell 2 — Create a dummy baseline pipeline

```python


dummy_pipeline = build_model_pipeline(
    model=DummyClassifier(
        strategy="prior"
    ),
    scale_numeric=False,
    add_engineered_features=False,
    imbalance_strategy="none"
)

print("Pipeline steps:")
print(list(dummy_pipeline.named_steps.keys()))
```

##### Cell 3 — Evaluate using five-fold cross-validation

```python

# - Evaluation

dummy_results = evaluate_cv(
    dummy_pipeline,
    X_train,
    y_train
)

display(dummy_results)
```
